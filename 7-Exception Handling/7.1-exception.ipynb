Understanding Exception

Exception handling in Python allows you to handle erroes gracefully and take corrective actions without stopping the excution of the program. In this section i will learn about the basics of exceptions , including how to use try,  exepct , else and finally blocks  

## What Are Exception? 

Exceptions are events that disrupt the normal flow of a program. They occur when an error is encounted during program exexution.

```python
try:
    num=int(input("Enter a number"))
    result=10/num
except ValueError:
    print("This is not a valid number")
except ZeroDivisionError:
    print("Enter denominator greater than 0")
except Exception as ex:
    print(ex)
```

```python
## try , except , else block

try:
    num=int(input("Enter a number"))
    result=10/num
except ValueError:
    print("This is not a valid number")
except ZeroDivisionError:
    print("Enter denominator greater than 0")
except Exception as ex:
    print(ex)
else:
    print(f"The result is {result}")

```

```python
## try, except, else and finally 

try:
    num=int(input("Enter a number"))
    result=10/num
except ValueError:
    print("This is not a valid number")
except ZeroDivisionError:
    print("Enter denominator greater than 0")
except Exception as ex:
    print(ex)
else:
    print(f"The result is {result}")
finally:
    print("Excetion Completed")
```
