## Estimator options

You can swap the default estimator with any function that turns a list of numbers into a single number.

| Function | Name | What it does |
|---|---|---|
| `np.mean` | Mean *(default)* | The average — adds all values and divides by the count |
| `np.median` | Median | The middle value when sorted; less thrown off by outliers |
| `np.sum` | Sum | Adds all values together |
| `len` | Count | Counts how many observations exist |
| `np.std` | Standard deviation | Measures how spread out the values are |
| `np.min` | Minimum | The smallest value |
| `np.max` | Maximum | The largest value |

# Box plot & violin plot

Both plots are used to show how data is **distributed** across categories.

---

### Box plot (box-and-whisker plot)

Shows the spread of data in a compact, comparable format — great for
comparing multiple groups side by side.

| Part | What it means |
|---|---|
| **Box** | The middle 50% of the data (Q1 to Q3, called the IQR) |
| **Line inside box** | The median — the exact middle value |
| **Whiskers** | Extend to the min and max values, excluding outliers |
| **Dots beyond whiskers** | Outliers — values unusually far from the rest |

---

### Violin plot

A violin plot is like a box plot, but the **width of the shape** shows you
where most values are concentrated — wider means more data points at that level.

- Shows everything a box plot shows (median, IQR, range)
- Also reveals the full **shape** of the distribution
- Useful for spotting skewness or multiple peaks in the data

> **Key difference:** A box plot tells you *where* the data sits;
> a violin plot also shows *how much* data sits there.

````python
sns.violinplot(x = 'day',y='total_bill',data=df)
````

## Strip plot & swarm plot

Both plots show **every individual data point** grouped by category —
unlike box or violin plots which summarize the data.

---

### Strip plot

A scatterplot where one axis is a category. Each dot = one observation.

- Simple and fast to draw
- Points can **overlap** when there is a lot of data, making it hard to read
- Works well **combined with** a box or violin plot to show raw data on top

---

### Swarm plot

Same as a strip plot, but points are **nudged sideways** so they never overlap.

- Every point is clearly visible
- Gives a more accurate picture of where data clusters
- Best for **small to medium datasets** — gets slow and crowded with too many points

---

### Quick comparison

| | Strip plot | Swarm plot |
|---|---|---|
| Points overlap? | Yes | No |
| Shows all data? | Yes | Yes |
| Works with large data? | Yes | Not ideal |
| Pairs with box/violin? | Great fit | Great fit |

# Categorical Plots — Summary

Categorical plots are used to **visualize the relationship between a numerical variable and a categorical variable**.
Each plot type gives a different view of the same data, so choosing the right one depends on what you want to show.

---

## Plot Types at a Glance

| Plot | What it shows | Best for |
|---|---|---|
| `boxplot` | Median, quartiles, and outliers | Comparing spread across groups |
| `violinplot` | Full shape of the distribution | Seeing skewness or multiple peaks |
| `stripplot` | Every individual data point | Showing raw data alongside other plots |
| `swarmplot` | Every point, no overlaps | Small datasets where each point matters |
| `barplot` | Average (mean) value per category | Comparing averages with confidence intervals |
| `countplot` | Number of observations per category | Counting how many rows fall in each group |

---

## 1. Box Plot

Shows how data is spread using five key numbers: minimum, Q1, median, Q3, and maximum.

- The **box** covers the middle 50% of the data (IQR)
- The **line** inside the box is the median
- The **whiskers** extend to the min/max (excluding outliers)
- **Dots** beyond the whiskers are outliers

```python
sns.boxplot(data=df, x="category", y="value")
```

---

## 2. Violin Plot

Like a box plot, but the **width** of the shape shows where most values are concentrated.

- Wider = more data points at that value
- Shows the full distribution shape, not just summary stats
- Can reveal bimodal (two-peak) distributions that a box plot would hide

```python
sns.violinplot(data=df, x="category", y="value")
```

---

## 3. Strip Plot

A scatterplot where one axis is a category. Each dot represents one observation.

- Simple and quick to draw
- Points **can overlap** when data is dense
- Great as a layer on top of a box or violin plot

```python
sns.stripplot(data=df, x="category", y="value")
```

---

## 4. Swarm Plot

Same as a strip plot, but points are **nudged sideways** so they never overlap.

- Every point is clearly visible
- Better representation of the distribution than a strip plot
- Not ideal for very large datasets — gets slow and crowded

```python
sns.swarmplot(data=df, x="category", y="value")
```

---

## 5. Bar Plot

Shows the **mean (average)** of a numerical variable for each category, with error bars showing the confidence interval.

- The height of each bar = the average value
- The error bar = uncertainty around that average
- Easy to compare averages across groups at a glance

```python
sns.barplot(data=df, x="category", y="value")
```

---

## 6. Count Plot

Shows **how many observations** fall into each category. No numerical variable needed — just counts rows.

- Like a bar plot, but the y-axis is always the count
- Useful for understanding class balance or frequency

```python
sns.countplot(data=df, x="category")
```

---

## Choosing the Right Plot

```
Do you want to show raw data points?
├── Yes → stripplot or swarmplot
│         (swarm if you need no overlap)
└── No → Do you want the full distribution shape?
          ├── Yes → violinplot
          ├── Summaries only → boxplot
          └── Just averages or counts?
                    ├── Averages → barplot
                    └── Counts  → countplot
```

---

> **Tip:** You can layer plots together for richer insights.
> For example, a `violinplot` + `swarmplot` on the same axes shows
> both the distribution shape and every individual data point.
