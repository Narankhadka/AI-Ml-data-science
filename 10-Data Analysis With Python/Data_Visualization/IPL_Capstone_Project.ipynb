# IPL 2022 Capstone Project

The **Indian Premier League (IPL)** is a professional T20 cricket league in India, featuring franchises representing cities. This project explores IPL 2022 match-level data to derive meaningful insights and understand match outcomes, player performances, and team dynamics.

---

## Dataset Overview

Below are the key columns focused on for meaningful insights in this project.

| Column Name | Variable Type | Description |
|---|---|---|
| `date` | String | Date of the match |
| `venue` | String | Stadium / ground where the match was played |
| `stage` | String | Stage of the tournament (e.g., Group Stage, Playoffs) |
| `team1` | String | Name of the first team |
| `team2` | String | Name of the second team |
| `toss_winner` | String | Team that won the toss |
| `toss_decision` | String | Decision made after winning the toss (bat/field) |
| `first_ings_score` | Integer | Total runs scored in the first innings |
| `second_ings_score` | Integer | Total runs scored in the second innings |
| `match_winner` | String | Team that won the match |
| `won_by` | String | Mode of victory (runs/wickets) |
| `margin` | Integer | Winning margin (runs or wickets) |
| `player_of_the_match` | String | Best performing player of the match |
| `top_scorer` | String | Highest run-scorer in the match |
| `highscore` | Integer | Highest individual score in the match |
| `best_bowling` | String | Best bowler in the match |
| `best_bowling_figure` | String | Bowling figures of the best bowler (e.g., 3/24) |

## Loading the Libraries and Dataset

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv("IPL.csv")
df.head()
```

## Some Basic question

1. Which team won the most matches?

```python
match_wins =df['match_winner'].value_counts()
```

```python
sns.barplot(y = match_wins.index,x=match_wins.values,palette ='viridis')
plt.title('Most match win by team')
```

2. Toss Decisions Trends

```python
sns.countplot(x=df['toss_decision'],palette='rainbow')
plt.title("Toss Decision Trends")
```

3. Toss Winner vs Match Winner

```python
df.head()
```

```python
count = df[df['toss_winner'] == df['match_winner']]['match_id'].count()
percentage =(count*100)/df.shape[0]
percentage.round(2)
```

4. How do teams win ? (Runs vs Wickets)

```python
df.head()
```

```python
sns.countplot(x=df['won_by'])
plt.title("Won by")
```

## Key Player Performances

```python
df.head()
```

1. Most "Player of the Match" Awards

```python
count = df['player_of_the_match'].value_counts().head(10)
count
```

```python
sns.barplot(x= count.values,y=count.index,palette='mako')
plt.title("Top 10 players man of the match")
```

2. Top Scores

```python
df.head()
```

```python
high = df.groupby('top_scorer')['highscore'].sum().sort_values(ascending=False).head(2)
high
```

```python
high.plot(kind='barh')
```

3 . Best Bowling Figures

```python
df.head()
```

```python
df['highest_wickets'] = df['best_bowling_figure'].apply(lambda x :x.split('--')[0])
df['highest_wickets'] = df['highest_wickets'].astype(int)
top_bowlers = df.groupby('best_bowling')['highest_wickets'].sum().sort_values(ascending=False).head(10)
top_bowlers.plot(kind = 'barh')
```
