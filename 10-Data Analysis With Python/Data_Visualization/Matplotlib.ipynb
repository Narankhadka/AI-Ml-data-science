## Matplotlib

### Matplotlib is a low level graph plotting library in python that serves as a visualization utility.

### Matplotlib was created by John D. Hunter.

### Matplotlib is open source and we can use it freely.

```python
import pandas as pad 
import numpy as np 
import matplotlib.pyplot as plt
```

## Matplotilb Basics

```python
x = np.linspace(0,5,11)
print(x)
y = x **3
print(y)
```

```python
plt.plot(x,y)
plt.title("Hello")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
```

## Subplots in Matplotlib

```python
plt.subplot(2, 2, 1)
plt.plot(x,y)


plt.subplot(2, 2, 2)
plt.plot(x,y)


plt.subplot(2, 2, 3)
plt.plot(y,x)

plt.subplot(2, 2, 4)
plt.plot(x,y)
```

## Object Oriented way

```python
fig = plt.figure(figsize=(10,8),dpi=200)
axis1 = fig.add_axes((0.1, 0.1, 0.3, 0.3))
axis1.plot(x,y)
axis2 = fig.add_axes((0.5, 0.5, 0.3, 0.3))
axis2.plot(x,y)
```

## Types of Plots in Matplotlib

### Basic Plots
| Plot | Function |
|------|----------|
| Line | `plt.plot()` |
| Scatter | `plt.scatter()` |
| Bar | `plt.bar()` / `plt.barh()` |
| Histogram | `plt.hist()` |
| Pie | `plt.pie()` |

### Statistical Plots
| Plot | Function |
|------|----------|
| Box Plot | `plt.boxplot()` |
| Violin | `plt.violinplot()` |
| Error Bar | `plt.errorbar()` |

### 2D/Grid Plots
| Plot | Function |
|------|----------|
| Heatmap | `plt.imshow()` |
| Contour | `plt.contour()` / `plt.contourf()` |
| 2D Histogram | `plt.hist2d()` |

### 3D Plots *(via `mpl_toolkits`)*
| Plot | Function |
|------|----------|
| 3D Line/Scatter | `ax.plot3D()` / `ax.scatter3D()` |
| 3D Surface | `ax.plot_surface()` |
| 3D Bar | `ax.bar3d()` |

### Other
| Plot | Function |
|------|----------|
| Stem | `plt.stem()` |
| Step | `plt.step()` |
| Stack Plot | `plt.stackplot()` |
| Quiver (arrows) | `plt.quiver()` |
| Polar | `plt.polar()` |

> Most plots are also accessible via the **OO interface**: `ax.plot()`, `ax.scatter()`, etc.
