# Regression Plots — Summary

Regression plots are used to **visualize the relationship between two numerical variables**
and fit a **regression line** to show the trend. They help you understand how one variable
changes as the other increases or decreases.

---

## Plot Types at a Glance

| Plot | What it shows | Best for |
|---|---|---|
| `regplot` | Scatter plot + regression line | Simple relationship between two variables |
| `lmplot` | `regplot` with extra grouping support | Comparing regression lines across categories |
| `residplot` | Residuals (errors) of the regression | Checking how well the model fits |

---

## 1. regplot

The most basic regression plot — draws a **scatterplot** and fits a **linear regression line** through the data.
A shaded band around the line shows the **confidence interval**.

```python
sns.regplot(data=df, x="variable_1", y="variable_2")
```

---

## 2. lmplot

Works like `regplot` but supports **grouping by a categorical variable** using `hue`, `col`, or `row`.
This lets you compare regression lines across different groups in one view.

```python
sns.lmplot(data=df, x="variable_1", y="variable_2", hue="category")
```

---

## 3. residplot

Shows the **residuals** — the difference between the actual values and the predicted values from the regression line.

- Points scattered randomly around zero = good fit
- A pattern in the residuals = the model is missing something

```python
sns.residplot(data=df, x="variable_1", y="variable_2")
```

---

## Key Concepts

| Term | Simple meaning |
|---|---|
| **Regression line** | The best-fit line that summarizes the trend in the data |
| **Confidence interval** | The shaded area — shows the uncertainty around the line |
| **Residual** | The gap between an actual data point and the predicted line |
| **Positive relationship** | As x goes up, y goes up |
| **Negative relationship** | As x goes up, y goes down |

---

## Choosing the Right Plot

```
Do you want to see the regression line?
├── Yes, simple two variables     → regplot
├── Yes, grouped by category      → lmplot
└── Want to check the model fit?  → residplot
```

---

> **Tip:** A regression plot does not prove that one variable **causes** the other to change —
> it only shows that they are **correlated**. Always think critically about the relationship.
