## Missing data 

```python
import numpy as np 
import pandas as pd
```

### Finding Missing Data

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df = pd.DataFrame(data)
df
```

```python
df.isna()
```

```python
df.isna().sum()
```

```python
df.isna().any()
```

#### Removing Missing Data

```python
df
```

```python
df.dropna()
```

```python
df
```

# Thresh 
### Key Mechanics:

Definition: It specifies the minimum number of non-NA values required to keep the row or column.

Logic: If a row has fewer than the thresh number of non-missing values, it is dropped.

Exclusivity: You cannot use thresh simultaneously with the how parameter

```python
df.dropna(thresh=3) ## at least 3  value should be there 
```

```python
df.dropna(thresh=4)
```
