## GroupBy

````python
data = {
    'Category': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
    'Store': ['S1', 'S1', 'S2', 'S2', 'S1', 'S2', 'S2', 'S1'],
    'Sales': [100, 200, 150, 250, 120, 180, 200, 300],
    'Quantity': [10, 15, 12, 18, 8, 20, 15, 25],
    'Date': pd.date_range('2023-01-01', periods=8)
}
df = pd.DataFrame(data)
df
````

### Group by Cagegory and calculate the sum of Sales

````python
cat =df.groupby('Category') [ 'Sales'].sum()
cat
````

### Group by Store and calculate the sum of Sales

````python
st= df.groupby('Store') ['Sales'].sum()
st
````

### Group by multiple columns
### Group by Category and store

````python
cat =  df.groupby(['Category','Store']) ['Sales'].sum()
cat
````

## Aggregation 
#### It means combining multiple values into a single summary result. It’s commonly used when you want statistics like sum, mean, count, min, max, etc., from your dataset.

## Basic Aggregation Methods

```python
# -------------------------------
# Basic Aggregation Methods
# -------------------------------

df['Marks'].sum()      # Returns total of Marks
df['Marks'].mean()     # Returns average Marks
df['Marks'].count()    # Counts non-null values
df['Marks'].min()      # Returns minimum value
df['Marks'].max()      # Returns maximum value
df['Marks'].median()   # Returns median value
df['Marks'].std()      # Returns standard deviation

````python
mean =df['Sales'].mean()
print(mean)
````

````python
# 'mode' does not work directly inside .agg() because it can return multiple values
result = df['Sales'].agg(['mean', 'min', 'max'])
result['mode'] = df['Sales'].mode().iloc[0]  # first mode
print(result)
````
