### Standard Library Overview 

Python's Standard Library is a vast collection of modules and packages that come bundled with Python , providing a wide range of functionalilites out of the box.  Here is an overview of some of the most commonly used modules and packages in th Python Standard Library.

```python
import array
arr=array.array('i',[1,2,3,4,5])
print(arr)
```

```python
import math
print(math.sqrt(16))
print(math.pi)
```

```python
## Random 
import random
print(random.randint(1,10))
print(random.choice(["apple","banana","cheery"]))
```

```python
## Data Serialization 

import json
data={'name':'naran','age':25}

json_str=json.dumps(data)
print(json_str)

paresed_data=json.loads(json_str)
print(paresed_data)
print(type(paresed_data))
```

```python
## cvs
import csv
with open('example.csv',mode='w',newline='')as file: 
    writer=csv.writer(file)
    writer.writerow(['name','age'])
    writer.writerow(['naran',12])

with open('example.csv',mode='r') as file:
    reader=csv.reader(file)
    for row in reader:
        print(row)
```

```python
## Date time 
from datetime import datetime, timedelta

now=datetime.now()
print(now)
yesterday=now-timedelta(days=1)
print(yesterday)
```

```python
## Time 
import time 
print(time.time())
time.sleep(2)
print(time.time())
```

```python
## Regular Expression 
import re 
pattern=r'\d+'
text='there are 123 apples 456'
match=re.search(pattern,text)
print(match.group())
```
