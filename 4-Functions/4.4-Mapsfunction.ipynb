The map() Function in python 

The map() function applies a given function to all items in an input list(or any other iterable) and returns a map objects(an iterator). This is particularly useful for transforming data in a list comprehensively 

```python
def squre(x):
    return x*x

squre(10)
```

```python
numbers=[1,2,3,4,5,6,7,8]
print(list(map(squre,numbers)))
```

```python
## Lambda function with map 
numbers=[1,2,3,4,5,6,7,8]
list(map(lambda x:x*x,numbers))
```

```python
### Map multiple iterables 

numbers1=[1,2,3]
numbers2=[4,5,6]
added_numbers=list(map(lambda x,y:x+y,numbers1,numbers2))
print(added_numbers)
```

```python
words=["apple","banana","cherry"]
upper_word=list(map(str.upper,words))
print(upper_word)
```

```python
def get_name(person):
    return person['name']

people=[
    {'name':'naran','age':31 },
    {'name':'ram','age':31}
]
list(map(get_name,people))
```

### Conclusion

The map() function is a powerful tool for applying transformations to iterable data structures. it can be used with regular functions, lambda functions, and even multiple itereable, providing a versatile approach to data processing in python. By understading and utilizing map() , we can write more efficient and readable code.
