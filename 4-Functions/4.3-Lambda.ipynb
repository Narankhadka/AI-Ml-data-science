Lambda function in python   

Lambda functions are small anonymous functions defined using tha lambda keyword. They can have any numbers of arguments but only one expression. They are commonly used for short operations or as arguments to higher-order functions

```python
#Syntax 
lambda arguments: expression
```

```python
def addition(a,b):
    return a+b
```

```python
addition(3,2)
```

```python
add=lambda a,b:a+b
type(add)
print(add(5,6))
```

```python
## Even 
def even(num):
    if num%2==0:
        return True

even(39)
```

```python
even1=lambda num:num%2==0 or False
even1(31)
```

```python
def addition(x,y,z):
    return x+y+z


addition(2,3,4)
```

```python
addition1=lambda x,y,z:x+y+z
addition1(120,22,10)
```

```python
##Map()- applies a fucntion to all items in a list 

numbers=[1,2,3,4,5,6,6,778,8,5,3]
def square(number):
    return number**2

square(2)
```

```python
list(map(lambda x:x**2,numbers))
```
