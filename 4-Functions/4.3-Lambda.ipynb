Lambda function in python   

Lambda functions are small anonymous functions defined using tha lambda keyword. They can have any numbers of arguments but only one expression. They are commonly used for short operations or as arguments to higher-order functions

```python
# Syntax
lambda argument:expression
```

```python
def addtiona(a,b):
    return a+b
```

```python
addtiona(2,4)
```

```python
additional=lambda a,b:a+b
print(type(additional))
print(additional(3,5))
```

```python
def even(num):
    if num%2==0:
        return True
even(24)
    
```

```python
even1=lambda num: num%2==0
print(even1(12))
```

```python
add=lambda x,y,z:x+y+z
print(add(1,2,3))
```

```python
## map() - applies a function to all items in a list
numbers=[1,2,3,4,5,6]
def squre(num):
    return num**2
squre(2)
```

```python
list(map(lambda x:x**2 , numbers))
```
