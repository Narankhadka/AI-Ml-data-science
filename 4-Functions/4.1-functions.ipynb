```python
##Default Parameters
def greet(name="guest"):
    print(f"Hello {name} Welcome to the paradise")

greet()
greet("naran")
```

```python
##Variable length Arguments


##Positional and keywords arguments

##Positional arguments
def print_numbers(*args):
    for number in args:
        print(number)
```

```python
print_numbers(2,2,3,4,5,3,2,23,4,55,"naran")
```

```python
##Keywords arguments
def print_details(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(name="naran", age="32",country="nepal")
```

```python
##Both 

##Keywords arguments
def print_detailswithargs(*args,**kwargs):
    for val in args:
        print(f"Positional argument :{val}")
    for key,value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_detailswithargs(1,3,3,4,5,name="naran", age="32",country="nepal")
```
