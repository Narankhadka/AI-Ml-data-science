```python
## Variable length arguments 
## Positional and keywords arguments 

## positional arguments

def print_numbers(*naran):
    for number in naran:
        print(number,end=" ")
```

```python
print_numbers(1,2,3,4,5,6,7,8,"narankhadka")
```

```python
### keyword arguments 

def print_details(**kwargs):
    for key , value in kwargs.items():
        print(f"{key}:{value}")

```

```python
print_details(name="naran",age="233",city="nepal")
```

```python
def print_details(*args,**kwargs):
    for val in args:
        print(f"Positional argument{val}",end=" ")
    for key , value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(1,2,3,4,18,"narankhadka",name="naran",age="233",city="nepal")
```
