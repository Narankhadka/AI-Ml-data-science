# Introduction

### Machine learning is a branch of artificial intelligence that enables algorithms to uncover hidden patterns within datasets. It allows them to predict new, similar data without explicit programming for each task

# Types of Machine Learning

Machine Learning algorithms can be broadly classified into three main categories based on how they learn from data and interact with their environment.

---

## 1. Supervised Learning

Supervised Learning involves training a model using **labeled data**, where both input and output values are provided.

### Key Features
- Learns from input-output pairs
- Predicts outcomes for new, unseen data
- Establishes a mapping between inputs and outputs

### Common Applications
- Classification
- Regression
- Forecasting

---

## 2. Unsupervised Learning

Unsupervised Learning works with **unlabeled data**, where the expected outputs are unknown.

### Key Features
- Identifies hidden patterns and relationships
- Discovers structures within datasets
- Groups similar data points automatically

### Common Applications
- Clustering
- Dimensionality Reduction
- Anomaly Detection

---

## 3. Reinforcement Learning

Reinforcement Learning is based on an **agent interacting with an environment** and learning through rewards and penalties.

### Key Features
- Learns through trial and error
- Receives feedback in the form of rewards or penalties
- Aims to maximize long-term rewards

### Common Applications
- Robotics
- Autonomous Systems
- Strategic Game Playing

---

# EDA (Exploratory Data Analysis)

EDA stands for **Exploratory Data Analysis**. It is the process of analyzing, visualizing, and understanding data before building a machine learning model.

The main goal of EDA is to discover patterns, identify relationships, detect anomalies, and gain insights from the dataset.
