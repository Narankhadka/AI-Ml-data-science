## SQL And SQLite

SQL(Structured Query Language) is a standard language for managing and manipulating relational databses. SQLite is a self-contained, serverless, and zero-configuration database engine that is widely used for embedded database systems. 

```python
import sqlite3
```

```python
## Connect to an SQLite database
connection = sqlite3.connect('example.db')
connection
```

```python
cursor = connection.cursor()
```

```python
## Create a Table
cursor.execute('''
create table if Not Exists employee(
    id Integer Primary key,
    name Text Not Null,
    age Integer, 
    department text
)
    ''')
## Commit the changes
connection.commit()
```

```python
## Insert the data in sqlite table 
cursor.execute('''
Insert Into employee(name,age,department)
values('naran',12,'data engineer')
               ''')
connection.commit()
```

```python
## Query the data from the table 
cursor.execute('select * from employee')
rows=cursor.fetchall()

## print the queried data 
for row in rows :
    print(row)
```

```python
## Update the data in the table 
cursor.execute(
    '''
    UPDATE employee
    set age =30 
    where name = "naran"
    '''
)
connection.commit()
```
