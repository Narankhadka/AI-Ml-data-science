#### SETS
Sets are a built-in data type in python used to store collections of unique items. They are unordered, meaning that the elements do not follow specific order, and they do not allow duplicate elements. Sets are useful for memebership tests, eliminating duplicate entries, and performing mathemetical set operations like union,intersection,differece,and sysmmetric difference. 

```python
## how to create a set 
my_set={1,2,3,4,5}
print(my_set)
print(type(my_set))
```

```python
my_set=set([1,2,3,4,5,6,2,2,1,4,1])
print(my_set)
```

```python
## Basic Sets operation 
## Adding and removing elements
my_set.add(8)
print(my_set)
my_set.add(8)
print(my_set)
```

```python
## Remove the elements from a set
my_set.remove(2)
print(my_set)
```

```python
my_set.discard(11) ## without an error it remove the element 
print(my_set)
```

```python
## pop method 
removed_element=my_set.pop()
print(removed_element)
print(my_set)
```

```python
my_set.clear()
print(my_set)
```

```python
## Set Memebership test
my_set={1,2,3,4,5,6}
list=[1,2,3]
print(3 in my_set)
print(10 in my_set)
print(1 in list)
```

```python
## Mathemaical operations 
set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9}

###Union 
union_set=set1.union(set2)
print(union_set)
```

```python
## Intersection 
intersection_set=set1.intersection(set2) ## common element on both 
print(intersection_set)
```

```python
set1.intersection_update(set2)
print(set1)
```

```python
print(set1.difference(set2))
```

```python
## Symmentric Difference 
set1.symmetric_difference(set2) ## common element has been remove
```

```python
## Sets Methods
set1={1,2,3,4,5,6,7,8}
set2={1,2,3,4,5}
## is subset
print(set1.issubset(set2))
```

```python
print(set1.issuperset(set2))
```

```python
### counting unique words in text
text=" I am naran khadka , and how are you doing "
words=text.split()

##convert lits of word to set to get unique words 
unique_words=set(words)
print(unique_words)
print(len(unique_words))
```
