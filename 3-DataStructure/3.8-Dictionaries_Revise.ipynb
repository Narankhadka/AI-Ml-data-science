#### Introduction to Dictionaries

Dictionaries are unordered collections of items. They store data in key-values pairs. Key must be unique and immutable (eg Strings, numbers , or tuples ),while values can by any type. 

```python
##Creating Dictionries
empty_dict = {}
print(type(empty_dict))
empty_sets=set() 
print(type(empty_sets))
# Important: Empty curly braces {} make a dictionary in Python.
# To create an empty set, always use set().
```

```python
student={"name":"naran","age":23,"grade":24}
print(student)
print(type(student))
```

```python
## Error 
student={"name":"naran","name":23,"grade":24}
print(student)
# Note:
# Dictionary keys must be unique in Python.
# If duplicate keys are used, the last assigned value will overwrite the previous one.
# In this case, "name" appears twice, so the value "naran" is replaced by 23.
```

```python
## accessing Dictionary Elements 
student={"name":"naran","age":23,"grade":'A'}
print(student)
print(student['age']) 

## Acessing using get() method
print(student.get('name'))
print(student.get('age'))
print(student.get('grade'))
print(student.get('lastname')) ## if there is no key and we access that one that will return the none 
print(student.get('lastname',"Not available")) ## we can customize the return value 
```

```python
## Modifying Dictinary Elements 
## Dictionary are mutable , so you can add , update or delete elements
student["age"]=33 ## updated value for the key
student["address"]="Nepal" ## Added a new key and value 
print(student)
```

```python
del student['grade'] ## delete key and value pair 
print(student)
```

```python
## Dictinary Method 
keys = student.keys() ## get all the keys  
print(keys)
values=student.values() ## get all the values 
print(values)
items= student.items() ## get all key value paris 
print(items) 
print(student)
```

```python
## Shallow copy 

student_copy=student
print(student)
print(student_copy)

```

```python
student["name"]="manisha karki"
print(student)
print(student_copy)
```

```python
student_copy1=student.copy() ## shallow copy 
print(student_copy1)
print(student)
student["name"]="hi student name"
print(student)
print(student_copy1)
```

```python
## Iterating over Dictionaries 
## We can use loops to iterate over dictionaries , key , values or items 

## Iterating over keys 

for keys in student.keys():
    print(keys)
```

```python
### Iterative over the values 
for values in student.values():
    print(values)
```

```python
## Iterate over key values pairs 
for key , value in student.items(): 
    print(f"{key}:{value}")
```

```python
## Nested Disctionaries 
students={
    "student":{"name":"naran","age":23},
    "student2":{"name":"sarthak","age":22},
    "student3":{"name":"pravab","age":20},
}
print(students)
```

```python
## Access nested dictionaries elements 
print(students["stuent2"]["name"])
```

```python
## Iterating over nested dictonaries 
for student_id, student_info in students.items():
    print(f"{student_id}:{student_info}")
    for key , value in student_info.items():
        print(f"{key}:{value}")
```

```python
## Dictinary Comprehension 
squres={ z:z**2 for z in range(5)}
print(squres)
```

```python
## conditional dictinary comprehension 
evens={x:x**2 for x in range(10) if x%2==0}
print(evens)
```

```python
## Practical Examples 
## Uuse a Dictionary to count the freequency of elements in list 
numbers=[1,2,2,3,3,3,4,4,4,4]
frequency={}
for number in numbers:
    if number in frequency:
        frequency[number]+=1
    else:
        frequency[number]=1
print(frequency)
```

```python
## Marge 2 dictionaries into one 
dict1={"a":1,"b":2}
dict2={"b":3,"c":4}
merged_dict = {**dict1,**dict2}
print(merged_dict)
```
