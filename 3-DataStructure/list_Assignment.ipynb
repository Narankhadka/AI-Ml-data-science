```python
# Module 3: Data Structures Assignments
## Lesson 3.1: Lists
### Assignment 1: Creating and Accessing Lists

#Create a list of the first 20 positive integers. Print the list.

lst=[]
for i in range(21):
    lst.append(i)
print(lst)
```

```python
### Assignment 2: Accessing List Elements

#Print the first, middle, and last elements of the list created in Assignment 1.
length=len(lst)
print(lst[0],lst[length//2],lst[length-1])
```

```python
### Assignment 3: List Slicing

#Print the first five elements, the last five elements, and the elements from index 5 to 15 of the list created in Assignment 1.
print(lst[:5])
print(lst[5:16])
print(lst[15:])
```

```python
### Assignment 4: List Comprehensions

#Create a new list containing the squares of the first 10 positive integers using a list comprehension. Print the new list.
new_list=[num**2 for num in range(11)]
print(new_list)
```

```python
### Assignment 5: Filtering Lists

#Create a new list containing only the even numbers from the list created in Assignment 1 using a list comprehension. Print the new list.

lst1=[num for num in lst if num%2==0]
print(lst1)
```

```python
### Assignment 6: List Methods

#Create a list of random numbers and sort it in ascending and descending order. Remove the duplicates from the list and print the modified list.
number_list=[12, 45, 7, 89, 23, 56, 34, 90, 18, 67]
print("The random list",number_list)
number_list.sort()
print("The ascending sort of number list",number_list)
number_list.sort(reverse=True)
print("The descending sort of number list",number_list)
not_duplicate_list=[] 
[not_duplicate_list.append(i) for i in number_list if i not in not_duplicate_list]
print("the new list", not_duplicate_list)
```
