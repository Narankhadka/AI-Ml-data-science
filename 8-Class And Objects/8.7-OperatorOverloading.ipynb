## Operator Overloading
 Operator Overloading allows us to define the behavior of operators(+,-,*,etc.) for custom object. We achieve this by overriding specific magic methods in our class

```python
#### Common Operatios Overloading Magic Methods
'''
__add__(self,other): Adds two objects using + operatos. 
__sub__(self, other): Substracts two objexts using the - operatos
etc
'''
```

```python
class Vector:
    def __init__(self,x,y):
        self.x=x
        self.y=y  
    def __add__(self,other):
        return Vector(self.x+other.x,self.y+other.y)
    def __sub__(self,other):
        return Vector(self.x-other.x,self.y-other.y)
    def __mul__(self,other):
        return Vector(self.x*other.x,self.y*other.y)
    def __eq__(self,other):
        return Vector(self.x==other.x,self.y==other.y)
    def __repr__(self):
        return  f"Vector {self.x},{self.y}"

## Create objects of the Vector class 
v1= Vector(2,3)
v2=Vector(4,5)

print(v1+v2)
print(v1-v2)
print(v1*v2)

```
