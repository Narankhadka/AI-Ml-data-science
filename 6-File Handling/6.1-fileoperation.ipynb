File Operation Read and Write files 


File handling is a crucial part of any programming langauge. Python provides built-in functions and methods to read from and write to files, both text and binary. 


```python
### Read a whole file 

with open ('simple.txt','r') as file:
    content=file.read()
    print(content)
        
```

```python
### read a file line by line 
with open ('simple.txt','r')as file:
    for line in file:
        print(line.strip()) ## strip() removes the newline character
```

```python
### Writing a file(Overwriting) this is important
with open('simple.txt','w') as file:
    file.write('Hello naran khadka i am from the ipynb \n')
    file.write('this is new line in ipynb\n')
```

```python
## write a file(without Overwriting)
with open('simple.txt','a')as file:
    file.write("append operation and this donot remove the exiting content")
```

```python
### Writing a list of lines to a file 

lines=['First line \n','Second line\n','Third line \n']
with open ('simple.txt','a') as file:
    file.writelines(lines)
```

```python
### Binary files 

#Writing to a binary file
data=b'\x00\x01\x02\x03\x04'
with open('example.bin','wb') as file:
    file.write(data)
```

```python
with open('example.bin','rb') as file:
    content=file.read()
    print(content)
  
```

```python
### Read the content from source text file and write to a destination text file 
with open('simple.txt','r')as file:
    content = file.read()

with open('destination.txt','w') as file:
    file.write(content)
```

```python
## Read a text file and count the numbers of lines, words, and characters.
#counting lines, words, and characters in a text file 

def count_text(file_path):
    with open(file_path,'r') as file:
        lines=file.readline()
        line_count=len(lines)
        word_count=sum(len(line.split()) for line in lines)
        char_count=sum(len(line) for line in lines)
    return line_count,word_count,char_count

file_path='simple.txt'
lines,words,characters=count_text(file_path)
print(f"Lines:{lines},words:{words},characters:{characters}")
```

The w+ mode in python is used to open file for both reading and writing. If the files doesnot exist, it will be created. if the files exits, its content is truncated(i.e, the file is overwritten)

```python
### Writing and then reading a files 
with open('simple.txt','w+') as file: 
    file.write("hello world and i am from w+")
    file.write("this is a new line \n")

    ## Move the files cursor to the begining 
    file.seek(0)
    # Read the content of the files 
    content=file.read()
    print (content)
```
