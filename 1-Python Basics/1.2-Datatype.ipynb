- Data types are a classicifcation of data which tell the compiler or interpreter how the programmer intends to use the data
- They determine the type of oeprations that can be perfomed on the data , the value that the data can take and the amount of memory needed to store the data. 


2. Importanace of Data types in programming 

Explanation: 

- Data types ensure that data is stored in an efficient way.
- they help in performing correct operations on data 
- proper use of data types can prevent errors and bugs in the program.
