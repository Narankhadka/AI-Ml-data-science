1. Introduction to the Operators 

```python
## Arithmethic operations 
a = 10 
b = 5
add_result=a+b #addtion
sub_result=a-b #substraction
div_result=a/b #divition
mul_result=a*b #multiplication
floor_div_result=a//b #floor divition
modules_result=a%b #modules operation 

exponsent_result=a**b #Exponentiation

print(add_result)
print(sub_result)
print(div_result)
print(mul_result)
print(floor_div_result)
print(modules_result)
print(exponsent_result)

```
