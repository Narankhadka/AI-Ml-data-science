They are commonly used to add functionality to functions or method without modifying their actual codde. In this section i will lean basic of decorators, includeing how to create and use them.

```python
### Decorator 

def main_welcome(func):

    def sub_welcome_method():
        print("welcome to sub_welcome_method")
        func()
        print("please give me a gift when you come to my house ")
        
    return sub_welcome_method()
```

```python
def  name_introduction():
    print("this is an introduction ")

name_introduction()
```

```python
main_welcome(name_introduction)
```

```python
@main_welcome
def  name_introduction():
    print("this is an introduction ")
```

```python
def my_decorator(func):
    def wrapper():
        print("Something is happening before the function is called")
        func()
        print("Something is happening after the function is called")
    return wrapper
```

```python
@my_decorator
def say_hello():
    print("Hello!")
```

```python
say_hello()
```

```python
## Decorators with arguments

def repeat(n):
    def decorator(func):
        def wrapper(*arg,**kwargs):
            for _ in range(n):
                func(*arg,**kwargs)
        
        return wrapper
    return decorator
        
```

```python
@repeat(3)
def say_hello():
    print("hello")
```

```python
say_hello()
```
