## Decorators 

#### Decorators are a powerful and flexible features in Python that allows you to modify the behavior of a function or class method. They are commonly used to add functionality to functions or method without modifying their actual codde. In this section i will lean basic of decorators, includeing how to create and use them.

```python
def main_welcome(func):

    def sub_welcome_method():
        print("welcome to sub_welcome_method")
        func("Welcome everyone to this section")
        print("please give me a gift when you come to my house ")
        
    return sub_welcome_method()
```
