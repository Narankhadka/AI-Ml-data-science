```python
## While loop 
count =0
while count<5:
    print(count)
    count=count+1
```

```python
count =0 
while count%2==0:
    print(count)
    count=count+1
    
```

```python
##Loop control statements
## Break 
## The break statement exits the loop permaturely 

##break statement 
for i in range(10):
    if i==5:
        break
    print(i)
```

```python
## continue 
## THe contine statment skips the current iteration and continues with the next.

#displaying the odd number between the range of 0 to 10 
for i in range(10):
    if i%2==0:
        continue # the example of the continue 
    print(i)
```

```python
## Pass  in python 
## The pass statement is a null opeartions , it does nothing 

for i in range(5):
    if i==3:
        print("the number is ",i)
        pass
    print(i)

```

```python
## Nested loops 
## a loop inside a loop 

for i in range(3):
    for j in range(2):
        print(f"I:{i} and J:{j}")
```

```python
## Examples- Calucalte the sum of first N natural numbers using a while loop and for loop

n=200099999
count=1
sum=0
while(count<=n):
    sum=sum+count
    count=count+1

print("Sum of first 10 natural nubmber: " ,sum)
```

```python
n =10
sum=0
for i in range(n+1):
    sum=sum+i

print(f"The natural number of {n} is {sum}")
```

```python
## Example- Prime numbers between 1 and 100 

for num in range (1,5):
    for i in range (2, num):
        if num%i==0:
            break
    else:
        print(num)
        
        

        
```
