```python
## else
## The else statment executes a block of code if the condition in the if statement is False.

age=16
if age>=18:
    print("You are allowed to have a driving license")
else:
    print("You are a minor")
```

```python
## elif
## The elif statement allows you to check multiple conditions. It stands for "else if"

age =12


if age<13:
    print("You are a child")
elif age <18:
    print("You are a teenager")
else:
    print("You are an adult")
```

```python
## Nested Conditional Statements 

## You can place one or more if ,elif , or else statmenet inside another if , elif , or else statement to create nested conditional statement. 

## number even, odd , negative

num = int (input("Enter the number"))

if num>0:
    print("The number is posistive ")
    if num%2==0:
        print("The number is even")
    else:
        print("the number is odd")
else:
    print("The number is zero or  negative")
```

```python
## Practical Example 
## Determine if a year is a leap year using nested condition statement 
 
year=int(input("Enter the year"))

if(year%4==0):
    if year%100==0:
        if year%400==0:
            print(f"{year} is a leap year")
        else:
            print(f"{year} is not a leap year")
    else:
        print(f"{year} is a leap year")
else:
       print(f"{year} is not a leap year")
```
