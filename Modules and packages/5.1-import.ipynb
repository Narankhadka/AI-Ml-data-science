Importing modules in Python : Modules and packages

In python ,moduels and packages help organize and reuse code .

```python
import math
math.sqrt(16)
```

```python
from math import sqrt , pi
print(sqrt(16))
print(sqrt(25))
print(pi)
```

```python
import numpy as np
np.array([1,2,3,4,5])
```

```python
from math import *
sqrt(16)
```

```python
from package.maths import addtion as ad
ad(2,3)
```

```python
from package import maths
maths.addtion(2,4)
```
